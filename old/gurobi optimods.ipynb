```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np

# Create a new model
m = gp.Model("trigonometric_example")

# Define the variable x as a difference of d1 and d2
d1 = m.addVar(name="d1")
d2 = m.addVar(name="d2")
z = m.addVar(lb=-np.pi, ub=np.pi, name="z")  # Ensure z is within the range of cosine function

# Add constraint that defines z
m.addConstr(z == d1 - d2, name="z_definition")
```

```python
# Define the variable y for the cosine of z
y = m.addVar(name="y")

# Add the cosine constraint
m.addGenConstrCos(z, y, name="cosine_constraint")
```

```python
# Objective: Maximize y
m.setObjective(y, GRB.MAXIMIZE)

# Optimize the model
m.optimize()

# Print the results
if m.status == GRB.OPTIMAL:
    print(f"Optimal value of y: {y.X}")
    print(f"Values of d1: {d1.X}, d2: {d2.X}")
    print(f"Value of z (should be d1 - d2): {z.X}")
else:
    print("Optimization was not successful.")
```
